```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler , LabelEncoder
import pickle
```

```python
import pandas as pd

#Load the dataset
data= pd.read_csv("Customer Attrition.csv")
print(type(data))
data.head()
```

```python
#Preprocess the data
#Drop irrelevant features
print(data.columns)
data=data.drop(['RowNumber','CustomerId','Surname'],axis=1)
data.columns=data.columns.str.strip()
data.columns
data.head()
print(data.columns)
```

```python
from sklearn.preprocessing import LabelEncoder

# Encode categorical feature
print(data['Gender'].head(10))
label_encoder_Gender=LabelEncoder()
data['Gender']=label_encoder_Gender.fit_transform(data['Gender'])
print(data['Gender'].head(10))
data
```

```python
# Encoding geography feature with labelencoder will assign value
# this may consider as more priority on bigger number so will mo
from sklearn.preprocessing import OneHotEncoder
one_hot_encode_geo =OneHotEncoder()
geo_encoder= one_hot_encode_geo.fit_transform(data[['Geography']])
geo_encoder
```

```python
#convert Compressed sparse row sparse matrix to array
geo_encoder.toarray()
```

```python
one_hot_encode_geo.get_feature_names_out(['Geography'])
```

```python
geo_encoded_df=pd.DataFrame(geo_encoder.toarray(),
                            columns=one_hot_encode_geo.
                            get_feature_names_out
                            (['Geography']))
geo_encoded_df
```

```python
data=pd.concat([data.drop('Geography',axis=1),geo_encoded_df],axis=1)
print(data)
```

### Merging One-Hot Encoded Geography and Dropping Original Column

We will now concatenate the `geo_encoded_df` (which contains the one-hot encoded 'Geography' features) with our main `data` DataFrame. After the concatenation, we will drop the original 'Geography' column from the `data` DataFrame as it is no longer needed.

```python
data = pd.concat([data.drop('Geography', axis=1), geo_encoded_df], axis=1)
data
```
